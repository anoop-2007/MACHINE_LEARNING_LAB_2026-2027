# Customer Segmentation using K-Means Clustering
Unsupervised learning mini project. No dataset upload needed: the data is generated inside the notebook.

## 1. Import libraries

```python
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
from sklearn.cluster import KMeans
from sklearn.preprocessing import StandardScaler
from sklearn.decomposition import PCA
from sklearn.metrics import silhouette_score
```

## 2. Create the customer dataset
We generate 300 customers in 5 hidden groups. To use your own file, upload a CSV and use pd.read_csv instead.

```python
rng = np.random.default_rng(42)

def make_group(n, age, income, spend, visits):
    return pd.DataFrame({
        "Age": rng.normal(age, 4, n),
        "AnnualIncome_k": rng.normal(income, 6, n),
        "SpendingScore": rng.normal(spend, 7, n),
        "VisitsPerMonth": rng.normal(visits, 1.5, n),
    })

df = pd.concat([
    make_group(60, 24, 30, 80, 10),   # young, low income, high spenders
    make_group(60, 45, 85, 75, 8),    # affluent, high spenders
    make_group(60, 50, 80, 20, 2),    # affluent, low spenders
    make_group(60, 55, 28, 25, 3),    # older, low income, low spenders
    make_group(60, 33, 55, 50, 5),    # average customers
]).clip(lower=1).reset_index(drop=True)

print("Shape:", df.shape)
df.head()
```

```python
df.describe().round(2)
```

## 3. Scale the features
K-Means is distance based, so all features must be on the same scale.

```python
scaler = StandardScaler()
X = scaler.fit_transform(df)
```

## 4. Choose the number of clusters
We use the Elbow method and the Silhouette score. A higher silhouette score means better-separated clusters.

```python
ks = range(2, 11)
inertias, sil_scores = [], []

for k in ks:
    km = KMeans(n_clusters=k, n_init=10, random_state=42).fit(X)
    inertias.append(km.inertia_)
    sil_scores.append(silhouette_score(X, km.labels_))

best_k = list(ks)[int(np.argmax(sil_scores))]
print(f"Best k by silhouette score: {best_k} (score = {max(sil_scores):.3f})")

fig, ax = plt.subplots(1, 2, figsize=(11, 4))
ax[0].plot(ks, inertias, marker="o")
ax[0].set(title="Elbow Method", xlabel="k", ylabel="Inertia")
ax[1].plot(ks, sil_scores, marker="o", color="green")
ax[1].set(title="Silhouette Score", xlabel="k", ylabel="Score")
plt.tight_layout()
plt.show()
```

## 5. Train the final model

```python
model = KMeans(n_clusters=best_k, n_init=10, random_state=42)
df["Cluster"] = model.fit_predict(X)
df["Cluster"].value_counts().sort_index()
```

## 6. Profile each segment
The average values in each cluster tell us what type of customer it represents.

```python
profile = df.groupby("Cluster").mean().round(1)
profile["Customers"] = df["Cluster"].value_counts().sort_index()
profile
```

## 7. Visualise the clusters with PCA
PCA reduces the 4 features to 2 dimensions so the clusters can be plotted.

```python
pca = PCA(n_components=2)
pts = pca.fit_transform(X)
centers = pca.transform(model.cluster_centers_)

plt.figure(figsize=(7, 5))
plt.scatter(pts[:, 0], pts[:, 1], c=df["Cluster"], cmap="viridis", s=35)
plt.scatter(centers[:, 0], centers[:, 1], c="red", marker="X", s=200, label="Centroids")
plt.title("Customer Segments (PCA view)")
plt.xlabel("PC1")
plt.ylabel("PC2")
plt.legend()
plt.show()
```

## 8. Predict the segment of a new customer

```python
new_customer = pd.DataFrame([[28, 35, 78, 9]], columns=df.columns[:-1])
segment = model.predict(scaler.transform(new_customer))[0]
print(f"New customer {new_customer.values.tolist()[0]} -> Segment {segment}")
```
